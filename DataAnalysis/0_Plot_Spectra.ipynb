# Plot a Molecule by its Name


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
import seaborn as sns
%matplotlib tk
# "../results/simulated_spectra/YYYY0625_HHmmss_combined_HPC.csv"
# "../results/simulated_spectra/20250603_190357_generatedSpectra.csv"

file = "../spectra_simulator/20250716_162424_generatedSpectra.csv"
molecule = "2-Amino-4-chlorobenzoic acid"

def parse_array_string(array_str):
    try:
        array_str = array_str.strip('"[]')
        array_str = array_str.replace(';', '')
        array_str = array_str.replace('  ', ' ')
        array_str = ','.join(array_str.split())
        return np.fromstring(array_str, sep=',')
    except Exception as e:
        print("Error parsing array string:", e)
        return np.array([])

df = pd.read_csv(file)

df.dropna(subset=['spectrum'],inplace=True) #drop the ones it couldn't generate
target = df[df["Name"] == molecule].iloc[0]

intensities_raw = target["spectrum"][2:-2]
intensities_strings = intensities_raw.split(';')
intensities = [float(num_str) for num_str in intensities_strings]
ppm = parse_array_string(df.loc[df['Name'] == molecule, 'ppm'].values[0])
sns.set_theme(font_scale=1.5)
plt.figure(figsize=(10, 4))
plt.plot(ppm, intensities, label=molecule)
plt.xlabel('Chemical Shift (ppm)')
plt.ylabel('Intensity')
plt.title(f"Simulated Spectra: {molecule}")
plt.gca().invert_xaxis()
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
